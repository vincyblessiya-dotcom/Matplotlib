```python
questions = [
    {
        "question": "What is the output of this code? x = [1, 2, 3]; print(x[::-1])",
        "options": ["[1, 2, 3]", "[3, 2, 1]", "[2, 1, 3]", "Error"],
        "answer": "B"
    },
    {
        "question": "What is the output? a = [1, 2, 3]; b = a; b.append(4); print(a)",
        "options": ["[1, 2, 3]", "[1, 2, 3, 4]", "[4]", "Error"],
        "answer": "B"
    },
    {
        "question": "What is the output? print([x for x in range(10) if x % 2 == 0 and x > 4])",
        "options": ["[2, 4, 6, 8]", "[6, 8]", "[4, 6, 8]", "[5, 6, 7, 8, 9]"],
        "answer": "B"
    },
    {
        "question": "Which technique is most effective for reducing overfitting in a machine learning model?",
        "options": ["Increasing model complexity", "Removing the validation set", "Regularization", "Increasing training epochs indefinitely"],
        "answer": "C"
    },
    {
        "question": "In a neural network, what problem is Batch Normalization primarily used to help address during training?",
        "options": ["Exploding/unstable activations and slow training", "Increasing the number of classes", "Converting images into text", "Removing the need for an optimizer"],
        "answer": "A"
    }
]
```

```python
def show_instructions():
    print("\n===== QUIZ INSTRUCTIONS =====")
    print("1. The quiz contains multiple-choice questions.")
    print("2. Each question has four options: A, B, C, and D.")
    print("3. Enter the letter of your chosen answer.")
    print("4. You will get 1 point for each correct answer.")
    print("5. There is no negative marking.")
    
```

```python
def start_quiz():
    correct = 0
    incorrect = 0

    for i, q in enumerate(questions, start=1):
        print("\nQuestion", i, ":", q["question"])

        print("A.", q["options"][0])
        print("B.", q["options"][1])
        print("C.", q["options"][2])
        print("D.", q["options"][3])

        answer = input("Enter your answer: ").upper()

        if answer == q["answer"]:
            print("Correct answer!")
            correct += 1
        elif answer in ["A", "B", "C", "D"]:
            print("Incorrect answer!")
            incorrect += 1
        else:
            print("Invalid choice!")

    show_results(correct, incorrect)

    again = input("\nWould you like to play again? (yes/no): ").lower()

    if again == "yes":
        start_quiz()
    else:
        print("Returning to main menu...")
```

```python
def show_results(correct, incorrect):
    total = len(questions)
    percentage = (correct / total) * 100

    print("\n===== QUIZ RESULTS =====")
    print("Total Questions:", total)
    print("Correct Answers:", correct)
    print("Incorrect Answers:", incorrect)
    print("Final Score:", percentage, "%")

    if percentage == 100:
        print("Performance: Excellent!")
    elif percentage >= 60:
        print("Performance: Good job!")
    else:
        print("Performance: Keep practicing!")
```

```python
def main():
    while True:
        print("\n===== QUIZ APPLICATION =====")
        print("1. Start Quiz")
        print("2. View Quiz Instructions")
        print("3. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            start_quiz()
        elif choice == "2":
            show_instructions()
        elif choice == "3":
            print("Thank you for playing!")
            break
        else:
            print("Invalid choice!")
```

```python
main()
```
